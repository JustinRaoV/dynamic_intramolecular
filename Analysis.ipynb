```python
import time
import os
from datetime import datetime
from collections import deque
from datetime import datetime

import re
from collections import deque
import csv

def pre_do(txt):
    file = open(txt, encoding='utf-8')
    # 此处为第一个输入
    # 打开txt格式文件（此文件为初始文件，目前需要在word或其他软件改成txt格式，否则utf-8会乱码），默认编码模式为UTF-8
    text = file.readlines()  # 将文件每一行写入text，形成一个text列表。
    file.close()  # 关闭原文件，对原文件进行保护，防止错误操作
    result = open("group2_ZV_2.txt", "w")  # 以写入的方式创建一个新的文件名称叫做ctest1.txt（warning：使用前保证不存在此名称文件即可）
    for lines in text:
        if lines[0] in ['A', 'C', 'T', 'G']:
            result.write(lines)  # 此循环用来过滤，判断第一位是否为ACTG其中之一，若是，则保留，若不是，则跳过。
    result.close()  # 关闭文件
    return None
    # 此函数无返回值，仅进行操作，不要进行赋值操作！


def pro_ls(txt):  # 将结果文档中的数据转化成列表
    file = open(txt, encoding='utf-8')
    text = file.readlines()
    lists = [x.strip() for x in text]
    return lists


def ls_max(ls):  # 获得有最大可能的核苷酸（ACTG)。
    A, C, G, T = 0, 0, 0, 0
    for i in ls:  # 遍历列表，查看下一位那个最大
        try:
            if i[0] == 'A':
                A += 1
            elif i[0] == 'C':
                C += 1
            elif i[0] == 'G':
                G += 1
            elif i[0] == 'T':
                T += 1
            else:
                continue  # 防止出现空格等
        except:
            continue

    def judge_num(num):  # 此函数为局部函数，用来返回最大的核苷酸
        if A == num:
            return 'A'
        elif C == num:
            return 'C'
        elif G == num:
            return 'G'
        elif T == num:
            return 'T'

    ls = []

    ls.append(A)
    ls.append(T)
    ls.append(C)
    ls.append(G)
    ls.sort()  # 按升序排列
    if ls[-1] < 1:
        print("软件出BUG了！请尝试重新修改参数，或尝试再次运行，下方运行结果不可靠")
        return None
    else:
        return judge_num(ls[-1]), ls[-1]  # 返回最大值


# 实现自动查找
class fastq(object):  # 定义一个fastq类对象
    def __init__(self):
        self.seq = None  # fastq实际序列
        self.start_seq = None  # fastq刚开始的序列
        self.remember_seq = None  # fastq的结果序列

    def next(self):
        data = []  # 核心步骤，实现向下查找一位
        ans = []
        for i in pro_lst:
            if self.start_seq in i:
                zls = i.split(self.start_seq)[1:]
                while '' in zls:
                    zls.remove('')
                data += zls
            else:
                continue
        for j in range(len(data) - 1):
            if data[j]:
                ans.append(data[j][0])
            else:
                continue
        max_ac, count = ls_max(ans)
        print(
            f"下一位中出现次数最多的元素是：{max_ac} ，总出现次数：{count}")
        newlist = list(filter(lambda x: x != max_ac, ans))
        
        if newlist:
            max_at, count = ls_max(newlist)
            print(
                f"下一位出现次数第二多的元素是：{max_at} ，总出现次数：{count}")
        newlist = list(filter(lambda x: x != max_at, newlist))
        
        if newlist:
            max_at, count = ls_max(newlist)
            print(
                f"下一位出现次数第三多的元素是：{max_at} ，总出现次数：{count}")
        newlist = list(filter(lambda x: x != max_at, newlist))
        
        
        if newlist:
            max_at, count = ls_max(newlist)
            print(
                f"下一位出现次数第三多的元素是：{max_at} ，总出现次数：{count}")

        self.start_seq += max_ac
        self.remember_seq += max_ac
        if self.remember_seq[:10] == self.remember_seq[-10:]:
            print("已经结束了")

    def cut_first(self):  # 去掉头，方便检索更多数据
        self.start_seq = self.start_seq[1:]  # 此步骤不会影响实际序列
        print("当前检索序列", self.start_seq)

    def cut_end(self):  # 去掉尾，人工加尾，对检索序列和储存序列同时操作
        self.start_seq = self.start_seq[:-1]
        self.remember_seq = self.remember_seq[:-1]
        print("当前检索序列", self.start_seq)

    def add(self, a):  # 人工添加最后一位
        self.start_seq += a
        self.remember_seq += a
        print("当前检索序列", self.start_seq)

    def is_end(self):  #
        if self.remember_seq[:21] == self.remember_seq[-21:]:
            print("已经结束")
        # 给个标志
```
